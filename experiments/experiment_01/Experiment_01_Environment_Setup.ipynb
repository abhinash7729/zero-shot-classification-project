```python
import torch
import transformers

from transformers import (
    pipeline,
    AutoTokenizer,
    AutoModelForSequenceClassification
)

print("PyTorch Version:", torch.__version__)
print("Transformers Version:", transformers.__version__)

print("Libraries imported successfully!")
```

```python
if torch.cuda.is_available():
    print("GPU Available:", torch.cuda.get_device_name(0))
    print("CUDA Version:", torch.version.cuda)
else:
    print("GPU not available. Running on CPU.")

device = 0 if torch.cuda.is_available() else -1
print("Selected Device:", device)
```

```python
MODEL_NAME = "facebook/bart-large-mnli"

print("Selected Model:", MODEL_NAME)

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME
)

print("Tokenizer loaded successfully!")
print("BART model loaded successfully!")
```

```python
print("MODEL CONFIGURATION")
print("-" * 40)

print("Model:", MODEL_NAME)
print("Architecture:", model.config.architectures)
print("Number of Labels:", model.config.num_labels)
print("Label Mapping:", model.config.id2label)
print("Vocabulary Size:", tokenizer.vocab_size)
print("Maximum Position Embeddings:",
      model.config.max_position_embeddings)
```

```python
import json
import os

os.makedirs("results", exist_ok=True)

experiment_config = {
    "project": "Project 3 - Zero-Shot Classification",
    "experiment": "Experiment 01",
    "model": MODEL_NAME,
    "framework": "PyTorch",
    "library": "Hugging Face Transformers",
    "python_version": sys.version,
    "torch_version": torch.__version__,
    "transformers_version": transformers.__version__,
    "gpu_available": torch.cuda.is_available(),
    "device": (
        torch.cuda.get_device_name(0)
        if torch.cuda.is_available()
        else "CPU"
    ),
    "status": "Environment configured"
}

with open("results/experiment_01_config.json", "w") as file:
    json.dump(experiment_config, file, indent=4)

print("Experiment configuration saved!")
```
